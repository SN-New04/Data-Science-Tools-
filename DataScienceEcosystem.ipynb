### Arithmetic Expression Examples
#### Below are a few examples of evaluating arithmetic expressions in Python

```python
# Addition
result = 8 + 7
print("Addition:", result)  # Output: 15

# Multiplication
result = 9 * 7
print("Multiplication:", result)  # Output: 63

# Division
result = 100 / 20
print("Division:", result)  # Output: 5.0

# Exponentiatio125
result = 5 ** 3
print("Exponentiation:", result)  # Output: 125
```
